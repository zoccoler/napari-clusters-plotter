```python
import napari
from pathlib import Path
from skimage.io import imread
import pandas as pd

viewer = napari.Viewer()

# Using pathlib.Path.cwd() to get current working directory
current_dir = Path.cwd()
print(f"Current working directory: {current_dir}")

# Get the notebook's directory (same as current working directory in Jupyter)
notebook_dir = Path.cwd()

# Construct path to sample data
path = notebook_dir / "src" / "napari_clusters_plotter" / "sample_data" / "BBBC007_v1_images" / "A9"
print(f"Sample data path: {path}")
print(f"Path exists: {path.exists()}")

label_filename = path / "A9 p5d_labels.tif"
feature_filename = path / "A9 p5d_features.csv"

labels = imread(label_filename)
features = pd.read_csv(feature_filename)
translate = labels.shape[-1]
viewer.add_labels(
    labels,
    name=label_filename.stem,
    features=features,
)

label_filename = path / "A9 p7d_labels.tif"
feature_filename = path / "A9 p7d_features.csv"
labels = imread(label_filename)
features = pd.read_csv(feature_filename)

viewer.add_labels(
    labels,
    name=label_filename.stem,
    features=features,
    translate=(0, translate)
)
```

```python
for layer in viewer.layers:
    if hasattr(layer, 'features') and layer.features is not None:
        print(f"Layer: {layer.name}, Features: {layer.features.shape}")
        layer.features.insert(1, 'layer_name', layer.name)
        layer.features['layer_name'] = layer.features['layer_name'].astype('category')
        print(layer.features.head())
```

```python
viewer.layers.selection
```

```python
# if selection changed
df_list = []
for layer in list(viewer.layers.selection):
    if hasattr(layer, 'features') and layer.features is not None:
        print(f"Selected Layer: {layer.name}, Features: {layer.features.shape}")
        if "layer_name" not in layer.features.columns:
            layer.features.insert(1, 'layer_name', layer.name)
            layer.features['layer_name'] = layer.features['layer_name'].astype('category')
        df_list.append(layer.features)

table = pd.concat(df_list)
table_inner = pd.concat(df_list, join='inner')
```
